## I2. Muutujad ja väärtused

Muutuja on nimi, mille kaudu kasutad väärtust. Märk `=` omistab nimele väärtuse. Nimed on tõstutundlikud: `vanus` ja `Vanus` on eri nimed. Kommentaar algab märgiga `#`; Python jätab kommentaari täitmisel vahele.

| Tüüp | Näide | Tähendus |
| --- | --- | --- |
| `int` | `24` | Täisarv |
| `float` | `1.72` | Ujukomaarv; kümnendmärk on punkt |
| `str` | `"Mari"` | Sõne ehk tekst |
| `bool` | `True`, `False` | Tõeväärtus |
| `None` | `None` | Väärtuse puudumist tähistav objekt |

`print()` kuvab tulemuse. `type()` näitab väärtuse tüüpi. Funktsiooni argumendid kirjutatakse sulgudesse.

````python
nimi = "Mari"
vanus = 24
pikkus = 1.72
osaleb = True
print(nimi, vanus, pikkus, osaleb)
print(type(vanus))
````

````python
hind = 2.50
kogus = 3
print(f"{kogus} eset maksavad {hind * kogus:.2f} eurot.")
print(int("12") + 1)
print(len("Python"))
print(7 % 2)
````

## I3. List, sõnastik ja meetod

**List** (`list`) hoiab järjestatud väärtusi. Nurksulgudes olev indeks valib elemendi: esimese indeks on `0`. Lõik `arvud[1:3]` sisaldab elemente indeksitel 1 ja 2; lõppindeks ei kuulu lõiku.

**Meetod** on objektiga seotud funktsioon, mida kutsud punktiga, näiteks `arvud.append(12)`. See lisab listi lõppu väärtuse. `min()`, `max()` ja `sum()` leiavad vastavalt väikseima väärtuse, suurima väärtuse ja summa.

````python
arvud = [4, 7, 9, 10]
print(arvud[0])
print(arvud[1:3])
arvud.append(12)
print(arvud)
print(min(arvud), max(arvud), sum(arvud))
````

**Sõnastik** (`dict`) hoiab võtme ja väärtuse paare. Väärtuse saamiseks kasuta võtit, mitte järjekorranumbrit. Sõnastik säilitab võtmete lisamise järjekorra; sama võtme uus omistus asendab varasema väärtuse.

````python
osaleja = {"nimi": "Mari", "vanus": 24}
print(osaleja["nimi"])
osaleja["linn"] = "Tartu"
print(osaleja)
````

## I4. Võrdlused ja tingimuslause

`==` võrdleb väärtusi; `=` omistab väärtuse. Teised võrdlusmärgid on `!=`, `<`, `<=`, `>` ja `>=`. Võrdluse tulemus on `True` või `False`. `in` kontrollib kuuluvust kogumisse.

Tingimusi saab ühendada operaatoritega `and` (ja), `or` (või) ja `not` (eitus).

````python
vanus = 24
print(vanus >= 18 and vanus < 65)
print("Tartu" in ["Tallinn", "Tartu"])
````

`if` käivitab ploki, kui tingimus kehtib. `elif` kontrollib järgmist tingimust ainult siis, kui eelmine ei kehtinud. `else` käivitub, kui ükski eelnev tingimus ei kehtinud.

Ploki algust tähistab koolon. Ploki read on **taandega**: kasuta nelja tühikut. Sama taandega read kuuluvad samasse plokki. Ära sega tabulaatoreid ja tühikuid.

````python
temperatuur = 3
if temperatuur < 0:
    print("Alla nulli")
elif temperatuur == 0:
    print("Null kraadi")
else:
    print("Üle nulli")
print("Kontroll lõppes.")
````

## I5. Tsükkel

`for` läbib kogumi elemente ükshaaval. Tsükli muutuja saab igal sammul järgmise elemendi väärtuse. Taandega kood kordub iga elemendi jaoks.

Järgmine näide loendab negatiivseid mõõtmisi. Ennusta tulemust enne käivitamist.

````python
mootmised = [-2, 4, -1, 0, 5]
negatiivseid = 0
for mootmine in mootmised:
    if mootmine < 0:
        negatiivseid = negatiivseid + 1
print(negatiivseid)
````

## I6. Funktsioon ja tagastusväärtus

Funktsioon koondab korduva tegevuse. `def` määrab funktsiooni nime ja **parameetrid**. Funktsiooni väljakutsel antud väärtused on **argumendid**. Funktsiooni keha käivitub alles väljakutsel.

`return` tagastab tulemuse edasiseks kasutamiseks. `print()` ainult kuvab väärtuse; see ei asenda tagastamist. Kui funktsioonis pole täidetud `return`-lauset, on tagastusväärtus `None`.

````python
def eurodest_sentideks(eurod):
    """Teisenda täiseurod sentideks."""
    sendid = eurod * 100
    return sendid

tulemus = eurodest_sentideks(3)
print(tulemus)
print(tulemus + 50)
````

`sendid` on funktsiooni kohalik muutuja: väljaspool funktsiooni ei saa seda nime otse kasutada. Tulemus jõuab välja `return` kaudu. `if`- ja `for`-plokk ei loo samal viisil eraldi lokaalset skoopi.

Tulemust saab kontrollida väitega `assert`. Kui tingimus on tõene, ei kuvata midagi. Väära tingimuse korral tekib `AssertionError`. Kasuta näiteid, mille vastust tead ette.

````python
assert eurodest_sentideks(0) == 0
assert eurodest_sentideks(4) == 400
print("Kontrollid õnnestusid.")
````

## I7. Veateate lugemine

Veateate lõpurealt leiad vea tüübi ja selgituse. Vaata seejärel rida, kus viga tekkis. `NameError` tähendab sageli valesti kirjutatud või loomata nime. `TypeError` viitab sobimatule tüübile; `ValueError` sobimatule väärtusele.

Näiteks `"12" + 1` tekitab `TypeError`, sest sõnet ja täisarvu ei saa nii liita. Arvutamiseks teisenda sõne arvuks: `int("12") + 1`.

`try` sisaldab koodi, mis võib vea anda. `except` käsitleb nimetatud erindit. Püüa konkreetset erindit; kõigi vigade peitmine raskendab vea põhjuse leidmist.

````python
sisend = "kaksteist"
try:
    arv = int(sisend)
    print(arv + 1)
except ValueError:
    print(f"Väärtust {sisend!r} ei saa täisarvuks teisendada.")
````

## I8. Teegid ja failid

`import` toob kasutusse mooduli. `pathlib` kuulub Pythoni standardteeki; Pandas ja Seaborn on eraldi paigaldatavad teegid.

`Path` tähistab failiteed. Punktiga kutsutav meetod, näiteks `mkdir()`, teeb tegevuse. **Atribuut**, näiteks `fail.name`, annab objekti omaduse ilma väljakutse sulgudeta.

Faili avamisel kasuta `with`-plokki: see sulgeb faili ka vea korral. Režiim `"w"` loob faili või kirjutab olemasoleva sisu üle; `"r"` loeb faili. `encoding="utf-8"` sobib eestikeelse teksti salvestamiseks.

````python
from pathlib import Path

kaust = Path("valjund")
kaust.mkdir(exist_ok=True)
fail = kaust / "tervitus.txt"

with fail.open("w", encoding="utf-8") as tekstifail:
    tekstifail.write("Tere, Python!\n")

with fail.open("r", encoding="utf-8") as tekstifail:
    sisu = tekstifail.read()

print(sisu)
print("Faili nimi:", fail.name)
print("Asukoht:", fail.resolve())
````

## I9. Lisamaterjal: objekt ja klass

Listid, sõnastikud ja andmetabelid on objektid. **Klass** kirjeldab, millised andmed ja meetodid teatud tüüpi objektil on. Meetodite kasutamiseks ei pea oskama uusi klasse kirjutada.

Järgmises näites seab `__init__` objekti algandmed. `self` viitab objektile endale. Laius ja kõrgus on atribuudid ning `pindala()` on meetod.

````python
class Ristkulik:
    def __init__(self, laius, korgus):
        self.laius = laius
        self.korgus = korgus

    def pindala(self):
        return self.laius * self.korgus

kujund = Ristkulik(3, 4)
print(kujund.laius)
print(kujund.pindala())
````
