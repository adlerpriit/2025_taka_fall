## I2. Muutujad ja väärtused

Muutuja on nimi, mille kaudu kasutad väärtust. Märk `=` omistab nimele väärtuse. Nimedes eristatakse suur- ja väiketähti: `vanus` ja `Vanus` on eri nimed. Kommentaar algab märgiga `#`; Python jätab selle osa reast koodi käivitamisel vahele.

| Tüüp | Näide | Tähendus |
| --- | --- | --- |
| `int` | `24` | Täisarv |
| `float` | `1.72` | Ujukomaarv; kümnendmärk on punkt |
| `str` | `"Mari"` | Sõne ehk tekst |
| `bool` | `True`, `False` | Tõeväärtus |
| `NoneType` | `None` | Väärtuse puudumist tähistav objekt |

`print()` kuvab tulemuse. `type()` näitab väärtuse tüüpi. Funktsioonile antavad väärtused ehk argumendid kirjutatakse funktsiooni nime järele sulgudesse.

`print()`, `type()` ja `len()` on **sisseehitatud funktsioonid**: need on kohe kasutatavad, midagi pole vaja importida ega paigaldada. [Sisseehitatud funktsioonide teatmikust](https://docs.python.org/3.12/library/functions.html) leiad kogu loendi koos parameetrite ja näidetega. Vaata alustuseks `print`, `len`, `type` ja `round`; kõiki funktsioone ei ole vaja pähe õppida.

````python
# Muutujad ja eri tüüpi väärtused.
nimi = "Mari"
vanus = 24
pikkus = 1.72
osaleb = True

# Väärtuste kuvamine funktsiooniga print().
print(nimi, vanus, pikkus, osaleb)

# Väärtuse tüübi vaatamine funktsiooniga type().
print(type(vanus))
````

Ennusta järgmise koodilahtri tulemust. Käivita see kaks korda ja võrdle tulemusi. Seejärel käivita uuesti eelmine koodilahter, kus `vanus = 24`, ning selle järel järgmine lahter. Võrdle tulemust varasemaga.

````python
vanus = vanus + 1
print(vanus)
````

Omistamine võib kasutada varasemat väärtust: kõigepealt arvutatakse parem pool ja seejärel seotakse nimi uue väärtusega.

Arvulise muutuja väärtuse muutmiseks on ka lühivormid ehk **liitomistamised**. Näiteks `+=` liidab olemasolevale väärtusele paremal antud arvu ja salvestab tulemuse samasse muutujasse. Eelmise näite rea `vanus = vanus + 1` võid seega kirjutada lühemalt: `vanus += 1`.

| Lühivorm | Pikem kuju arvulise muutuja korral | Tegevus |
| --- | --- | --- |
| `arv += 2` | `arv = arv + 2` | Suurendab väärtust kahe võrra |
| `arv -= 2` | `arv = arv - 2` | Vähendab väärtust kahe võrra |
| `arv *= 2` | `arv = arv * 2` | Korrutab väärtuse kahega |
| `arv /= 2` | `arv = arv / 2` | Jagab väärtuse kahega |

Need käsud muudavad muutuja väärtust. Võrdlusmärke `==` ja `!=` vaatame peatükis [I4](#i4).

Arvutustes saad kasutada märke `+`, `-`, `*`, `/` ja `**` (astendamine). `%` annab jagamisjäägi. `len()` annab kogumi või sõne pikkuse. Väärtuse teisendamiseks teise tüüpi saad kasutada näiteks `int()`, `float()` ja `str()`.

**F-sõne** korral on sõne alustava jutumärgi ees täht `f`. Loogelistesse sulgudesse saad kirjutada muutuja või arvutuse, mille tulemus lisatakse teksti. Vorming `:.2f` kuvab arvu kahe komakohaga.

Dokumentatsioon: [arvud ja sõned](https://docs.python.org/3.12/tutorial/introduction.html#using-python-as-a-calculator), [f-sõned](https://docs.python.org/3.12/tutorial/inputoutput.html#formatted-string-literals).

````python
# F-sõne: arvutuse tulemuse lisamine teksti kahe komakohaga.
hind = 2.50
kogus = 3
print(f"{kogus} eset maksavad {hind * kogus:.2f} eurot.")

# Tüübi teisendamine: sõnest täisarvuks.
print(int("12") + 1)

# Sõne pikkuse leidmine.
print(len("Python"))

# Jagamisjäägi leidmine.
print(7 % 2)
````

## I3. Andmestruktuurid, funktsioonid ja meetodid

**List** (`list`) hoiab väärtusi kindlas järjekorras. Nurksulgudes olev indeks näitab elemendi asukohta listis: esimese elemendi indeks on `0`.

Mitme järjestikuse elemendi valimiseks kasuta **lõiku** (*slice*). Näiteks `arvud[1:3]` tagastab uue listi elementidega indeksitel 1 ja 2 ehk esialgse listi teise ja kolmanda elemendiga. Valik algab esimesest antud indeksist ja lõpeb enne teist: indeksiga 3 elementi kaasa ei võeta.

Listi saad anda argumendiks ka sisseehitatud funktsioonile: `min(arvud)`, `max(arvud)` ja `sum(arvud)` tagastavad vastavalt väikseima väärtuse, suurima väärtuse ja summa. Need ei muuda listi.

**Meetod** on objektiga seotud funktsioon. Meetodi väljakutses kirjuta objekti tähistava nime järele punkt, meetodi nimi ja sulud: `arvud.append(12)`. Listi meetod `append()` muudab olemasolevat listi, lisades lõppu väärtuse. See tagastab `None`, seega ära kirjuta `arvud = arvud.append(12)`.

Võrdle kirjapilti: `sum(arvud)` kutsub funktsiooni, `arvud.append(12)` listi meetodit. Kõik meetodid ei muuda objekti: näiteks sõne meetod `"Mari".upper()` tagastab uue sõne `"MARI"`.

Dokumentatsioon: [listi meetodid](https://docs.python.org/3.12/tutorial/datastructures.html#more-on-lists), [min, max ja sum](https://docs.python.org/3.12/library/functions.html).

````python
# Listi loomine ja esimese elemendi valimine indeksiga.
arvud = [4, 7, 9, 10]
print(arvud[0])

# Lõik: teise ja kolmanda elemendi valimine.
print(arvud[1:3])

# Meetod append(): olemasoleva listi muutmine.
arvud.append(12)
print(arvud)

# Sisseehitatud funktsioonid: väikseim väärtus, suurim väärtus ja summa.
print(min(arvud), max(arvud), sum(arvud))
````

**Sõnastik** (`dict`) hoiab võtme ja väärtuse paare. Väärtuse saamiseks kasuta võtit, mitte järjekorranumbrit. Sõnastik säilitab võtmete lisamise järjekorra; samale võtmele uue väärtuse määramine asendab varasema väärtuse.

Dokumentatsioon: [sõnastikud](https://docs.python.org/3.12/tutorial/datastructures.html#dictionaries).

````python
# Sõnastiku loomine ja väärtuse valimine võtme järgi.
osaleja = {"nimi": "Mari", "vanus": 24}
print(osaleja["nimi"])

# Uue võtme ja väärtuse paari lisamine.
osaleja["linn"] = "Tartu"
print(osaleja)
````

### Põhiliste andmestruktuuride ülevaade

Järgmisi tüüpe tasub koodi lugedes ära tunda. Tabeli viimane veerg näitab, kas olemasoleva kogumi elemente saab asendada, lisada või eemaldada.

| Tüüp | Näide | Järjekord ja valimine | Kas kogumit saab muuta? |
| --- | --- | --- | --- |
| List (`list`) | `[4, 7, 4]` | Järjestatud; valik indeksiga | Jah |
| Ennik (`tuple`) | `(4, 7)` | Järjestatud; valik indeksiga | Ei |
| Sõnastik (`dict`) | `{"nimi": "Mari", "vanus": 24}` | Säilitab lisamise järjekorra; valik võtmega | Jah |
| Hulk (`set`) | `{"A", "T", "S"}` | Kordumatud elemendid; kindlat järjekorda ega indeksit pole | Jah |
| Muutumatu hulk (`frozenset`) | `frozenset({"A", "T"})` | Kordumatud elemendid; kindlat järjekorda ega indeksit pole | Ei |
| Täisarvujada (`range`) | `range(1, 4)` | Järjestatud; siin väärtused 1, 2 ja 3 | Ei |

Tühi `{}` on sõnastik; tühja hulga loob `set()`. Ühe elemendiga ennikus on koma: `(4,)`. Ennik sobib näiteks koordinaatide hoidmiseks või funktsioonist mitme väärtuse korraga tagastamiseks. Enniku elemente ei saa asendada, kuid selle sees olevat listi saab siiski muuta.

Hulk sobib kordumatute väärtuste leidmiseks ja kontrollimiseks, kas väärtus kuulub hulka. `range` kirjeldab täisarvude jada ilma kõiki arve listi salvestamata; kasutad seda tsüklite juures. Ka sõne on järjestatud ja muutumatu: `"Mari"[0]` annab `"M"`.

Dokumentatsioon: [ennikud](https://docs.python.org/3.12/tutorial/datastructures.html#tuples-and-sequences), [hulgad](https://docs.python.org/3.12/tutorial/datastructures.html#sets), [set ja frozenset](https://docs.python.org/3.12/library/stdtypes.html#set-types-set-frozenset), [range](https://docs.python.org/3.12/library/stdtypes.html#ranges).

````python
# Ennik: väärtuste hoidmine koos ja omistamine eri muutujatele.
koordinaadid = (58.38, 26.72)
laius, pikkus = koordinaadid  # Esimene väärtus omistatakse muutujale laius, teine muutujale pikkus.
print(laius, pikkus)

# Hulk: kordumatute väärtuste leidmine.
ravimikoodid = ["A", "T", "A", "S"]
unikaalsed = set(ravimikoodid)
print(len(unikaalsed))

# Hulka kuulumise kontrollimine.
print("A" in unikaalsed)

# Hulga väärtustest järjestatud listi loomine.
print(sorted(unikaalsed))  # sorted() tagastab järjestatud listi.
````

## I4. Võrdlused ja tingimuslause

`==` kontrollib, kas väärtused on võrdsed; `!=` kontrollib, kas need on erinevad. Üksik `=` omistab väärtuse. Teised võrdlusmärgid on `<`, `<=`, `>` ja `>=`. Võrdluse tulemus on `True` või `False`. `in` kontrollib, kas väärtus kuulub kogumisse.

Tingimusi saab ühendada operaatoritega `and` (ja), `or` (või) ja `not` (eitus).

````python
# Kahe võrdluse ühendamine operaatoriga and.
vanus = 24
print(vanus >= 18 and vanus < 65)

# Listi kuulumise kontroll operaatoriga in.
print("Tartu" in ["Tallinn", "Tartu"])
````

`if` käivitab ploki, kui tingimus kehtib. `elif` kontrollib järgmist tingimust ainult siis, kui eelmine ei kehtinud. `else` käivitub, kui ükski eelnev tingimus ei kehtinud.

Tingimuse rea lõpus on koolon. Sellele järgnevad ploki read kirjuta **taandega**, lisades rea algusesse neli tühikut. Plokk lõpeb seal, kus järgmise koodirea taane on väiksem. Ära sega tabulaatoreid ja tühikuid.

Dokumentatsioon: [võrdlused ja loogika](https://docs.python.org/3.12/library/stdtypes.html#boolean-operations-and-or-not), [if-laused](https://docs.python.org/3.12/tutorial/controlflow.html#if-statements).

````python
# Tingimuslause: käivitub ainult sobiva haru taandega kood.
temperatuur = 3
if temperatuur < 0:
    print("Alla nulli")
elif temperatuur == 0:
    print("Null kraadi")
else:
    print("Üle nulli")

# Taandeta rida jääb tingimuslausest välja ja käivitub iga haru järel.
print("Kontroll lõppes.")
````

## I5. Tsükkel

`for` läbib kogumi elemente ükshaaval. Tsükli muutuja saab igal sammul järgmise elemendi väärtuse. Tsükli sees olevat taandega koodi täidetakse iga elemendi jaoks.

Järgmine näide loendab negatiivseid mõõtetulemusi. Ennusta tulemust enne käivitamist.

````python
mootmised = [-2, 4, -1, 0, 5]
negatiivseid = 0
for mootmine in mootmised:
    if mootmine < 0:
        negatiivseid = negatiivseid + 1
print(negatiivseid)
````

## I6. Funktsioon ja tagastusväärtus

Funktsioon võimaldab tegevuse ühe korra kirja panna ja seda eri väärtustega korrata. Funktsiooni defineerimisel kirjuta `def` järele funktsiooni nimi ning sulgudesse **parameetrid**. Parameetrid on nimed, mille kaudu funktsioon kasutab talle antud väärtusi. Väljakutsel antud väärtusi nimetatakse **argumentideks**. Funktsiooni sees olev kood käivitub alles siis, kui funktsiooni välja kutsud.

`return` tagastab tulemuse edasiseks kasutamiseks. `print()` ainult kuvab väärtuse; see ei asenda tagastamist. Kui funktsiooni täitmine jõuab lõpuni ilma `return`-lauseta, tagastab funktsioon `None`.

Dokumentatsioon: [funktsiooni defineerimine](https://docs.python.org/3.12/tutorial/controlflow.html#defining-functions).

````python
# Funktsiooni defineerimine: parameeter ja tulemuse tagastamine.
def eurodest_sentideks(eurod):
    """Teisenda täiseurod sentideks."""
    sendid = eurod * 100
    return sendid

# Funktsiooni väljakutse ja tagastatud väärtuse kasutamine.
tulemus = eurodest_sentideks(3)
print(tulemus)
print(tulemus + 50)
````

`sendid` on funktsiooni kohalik muutuja: väljaspool funktsiooni ei saa seda nime otse kasutada. Funktsiooni arvutatud tulemuse saad kätte `return` abil. Seevastu `if`- või `for`-plokis loodud muutuja on kasutatav ka pärast ploki lõppu, kui muutuja loonud rida käivitati.

Tulemuse kontrollimiseks saad kasutada `assert`-lauset. Kui tingimus on tõene, ei kuvata midagi. Väära tingimuse korral tekib `AssertionError`. Kasuta näiteid, mille vastust tead ette.

Dokumentatsioon: [assert](https://docs.python.org/3.12/reference/simple_stmts.html#the-assert-statement).

````python
assert eurodest_sentideks(0) == 0
assert eurodest_sentideks(4) == 400
print("Kontrollid õnnestusid.")
````

## I7. Veateate lugemine

Veateate viimaselt realt leiad vea tüübi ja selgituse. Vaata seejärel rida, kus viga tekkis. `NameError` tähendab, et Python ei leia kasutatud nime. Näiteks võib muutuja nimi olla valesti kirjutatud või selle loonud lahter käivitamata. `TypeError` viitab sobimatule tüübile; `ValueError` sobimatule väärtusele.

Näiteks `"12" + 1` tekitab vea `TypeError`, sest sõnet ja täisarvu ei saa nii liita. Arvutamiseks teisenda sõne arvuks: `int("12") + 1`.

Koodi käivitamisel tekkivat viga nimetatakse **erindiks**. `try`-plokki kirjuta kood, mille puhul soovid võimalikku viga käsitleda. `except`-plokk käivitub siis, kui tekib seal nimetatud tüüpi erind. Nimeta oodatav veatüüp, näiteks `ValueError`; kõigi vigade peitmine raskendab vea põhjuse leidmist.

Dokumentatsioon: [vead ja erindid](https://docs.python.org/3.12/tutorial/errors.html).

````python
sisend = "kaksteist"

# Proovi teisendada sõne täisarvuks.
try:
    arv = int(sisend)
    print(arv + 1)

# Käsitle teisendamisel tekkivat ValueError-erindit.
except ValueError:
    print(f"Väärtust {sisend!r} ei saa täisarvuks teisendada.")
````

Kõike ei pea ise kirjutama. **Moodul** koondab Pythoni koodi, näiteks funktsioone ja klasse. **Teegiks** nimetatakse korduskasutatava koodi kogumit, mis võib koosneda mitmest moodulist ja paketist.

| Mida kasutad? | Näited | Mida on vaja teha? |
| --- | --- | --- |
| Sisseehitatud funktsioonid | `print()`, `len()`, `sum()` | Saab kohe kasutada |
| Pythoni standardteek | `math`, `pathlib`, `csv` | Pythoniga kaasas; vajalik moodul või nimi tuleb importida |
| Eraldi paigaldatavad teegid | Pandas, Seaborn, Matplotlib | Teek peab olema keskkonda paigaldatud; seejärel impordi |

**Paigaldamine** lisab teegi Pythoni keskkonda. **Importimine** teeb selle koodi käimasolevas programmis kasutatavaks. `import` ise teeki internetist alla ei laadi. Pärast kerneli taaskäivitamist käivita impordilahtrid uuesti.

`import math` teeb mooduli kättesaadavaks nimega `math`; `math.sqrt(9)` kutsub selle ruutjuurefunktsiooni. Punkt võib seega viidata ka moodulis olevale funktsioonile, mitte ainult objekti meetodile. `from pathlib import Path` toob moodulist `pathlib` kasutusse nime `Path`, mistõttu saad kirjutada `Path(...)`, mitte `pathlib.Path(...)`. `import pandas as pd` annab Pandasele selles vihikus lühinime `pd`.

Teekide paigaldamiseks järgi [alustamisjuhist](README.md). Seal kasutatav terminalikäsk `python -m pip install -r requirements.txt` paigaldab failis `requirements.txt` loetletud teegid. Colabis on Pandas ja Seaborn tavaliselt juba olemas, kuid importida tuleb need ikka. Kui tekib `ModuleNotFoundError`, kontrolli esmalt, millist Pythoni keskkonda kernel kasutab.

Dokumentatsioon: [moodulid ja import](https://docs.python.org/3.12/tutorial/modules.html), [standardteegi ülevaade](https://docs.python.org/3.12/library/index.html), [teekide paigaldamine](https://docs.python.org/3.12/tutorial/venv.html#managing-packages-with-pip).

````python
# Standardteegi mooduli importimine.
import math

# Moodulis oleva funktsiooni kutsumine.
print(math.sqrt(9))
````

### Failitee ja faili sisu

`Path` tähistab failiteed. Meetod `kaust.mkdir()` loob kausta. Objektil on ka **atribuudid** ehk omadused: näiteks `fail.name` annab failinime. Atribuudi väärtuse lugemisel väljakutse sulge ei kasutata.

Faili avamisel kasuta `with`-plokki: plokist väljumisel suletakse fail ka siis, kui tekkis viga. Režiim `"w"` loob faili või kirjutab olemasoleva sisu üle; `"r"` loeb faili. `encoding="utf-8"` sobib eestikeelse teksti salvestamiseks.

Faili saad kirjutada tuttava funktsiooniga `print()`. Argument `file=tekstifail` suunab väljundi ekraani asemel avatud faili. Siin on `file` parameetri nimi ja `tekstifail` viide `with`-plokis avatud failile.

Nagu ekraanile kuvamisel, saad anda `print()`-ile mitu väärtust: näiteks `print("Vanus:", 24, file=tekstifail)`. Vaikimisi eraldab `print()` väärtused tühikuga ja lisab lõppu reavahetuse. Eraldajat saad muuta parameetriga `sep`, rea lõppu parameetriga `end`.

Faili meetod `write()` sobib valmis sõne kirjutamiseks. Näiteks `tekstifail.write("Tere, Python!\n")` kirjutab sama teksti kui allolev `print()`-käsk. `write()` ise reavahetust ei lisa; siin tähistab seda `\n`.

Dokumentatsioon: [print](https://docs.python.org/3.12/library/functions.html#print), [Path](https://docs.python.org/3.12/library/pathlib.html#basic-use), [failide lugemine ja kirjutamine](https://docs.python.org/3.12/tutorial/inputoutput.html#reading-and-writing-files).

````python
from pathlib import Path

# Failitee koostamine ja kausta loomine.
kaust = Path("valjund")
kaust.mkdir(exist_ok=True)
fail = kaust / "tervitus.txt"

# Faili kirjutamine print() abil.
with fail.open("w", encoding="utf-8") as tekstifail:
    print("Tere, Python!", file=tekstifail)

# Faili sisu lugemine.
with fail.open("r", encoding="utf-8") as tekstifail:
    sisu = tekstifail.read()

# Loetud teksti kuvamine ilma lisareavahetuseta.
print(sisu, end="")  # Failist loetud tekst lõpeb juba reavahetusega.

# Atribuut name: failinime lugemine.
print("Faili nimi:", fail.name)

# Meetod resolve(): faili täieliku asukoha leidmine.
print("Asukoht:", fail.resolve())
````

## I9. Lisamaterjal: objekt ja klass

Listid, sõnastikud ja andmetabelid on objektid. **Klass** kirjeldab, millised andmed ja meetodid teatud tüüpi objektil on. Meetodite kasutamiseks ei pea oskama uusi klasse kirjutada.

Järgmises näites määrab `__init__` uue ristküliku laiuse ja kõrguse. `self` viitab objektile endale. Laius ja kõrgus on atribuudid ning `pindala()` on meetod.

Dokumentatsioon: [klassid ja objektid](https://docs.python.org/3.12/tutorial/classes.html#a-first-look-at-classes).

````python
# Klassi kirjeldus: objekti atribuudid ja meetod.
class Ristkulik:
    def __init__(self, laius, korgus):
        self.laius = laius
        self.korgus = korgus

    def pindala(self):
        return self.laius * self.korgus

# Objekti loomine selle klassi põhjal.
kujund = Ristkulik(3, 4)

# Atribuudi väärtuse lugemine.
print(kujund.laius)

# Meetodi kutsumine pindala arvutamiseks.
print(kujund.pindala())
````
